```python
gb = df.groupby("ParentEduc").agg({"MathScore":'mean',"ReadingScore":"mean","WritingScore":"mean"})
print(gb)
```

```python
sns.heatmap(gb, annot = True)
plt.show()
```

```python
# from the above chart we have concluded that parents education have a good impact in childrents good marks 
```

```python
gb1 = df.groupby("ParentMaritalStatus").agg({"MathScore":"mean","ReadingScore":"mean","WritingScore":"mean"})
print(gb1)
```

```python
sns.heatmap(gb1,annot = True)
plt.title("Relationship between parents marital status and students score")
plt.show()
```

```python
# from the above chart we have concluded that there is no/neglible impact due to their marital status
```
